```python
# Machine Learning Model
# Import machine learning libraries

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    GridSearchCV,
    cross_validate
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("Machine learning libraries imported successfully.")
```

```python
# Prepare input features and target

target = "Number of Doctors Visited"

X = df.drop(columns=[target])
y = df[target]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Check target distribution in the cleaned dataset

print("Target counts:")
print(y.value_counts().sort_index())

print("\nTarget percentages:")
print(
    y.value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
# Split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Identify categorical input columns

categorical_columns = X.columns.tolist()

print("Number of categorical columns:", len(categorical_columns))

print("\nCategorical columns:")

for column in categorical_columns:
    print("-", column)
```

```python
# Create the categorical preprocessing component

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_columns
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
# Create the baseline Decision Tree

decision_tree = DecisionTreeClassifier(
    random_state=42
)

print("Decision Tree created successfully.")
```

```python
# Combine preprocessing and model into one pipeline

pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", decision_tree)
    ]
)

print(pipeline)
```

```python
# Train the baseline Decision Tree

pipeline.fit(
    X_train,
    y_train
)

print("Baseline Decision Tree trained successfully.")
```

```python
# Generate baseline predictions

y_pred = pipeline.predict(
    X_test
)

print("Predictions generated successfully.")
```

```python
# Calculate baseline accuracy

baseline_accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Baseline Accuracy:", baseline_accuracy)
```

```python
# Calculate baseline evaluation metrics

baseline_precision = precision_score(
    y_test,
    y_pred,
    average="macro"
)

baseline_recall = recall_score(
    y_test,
    y_pred,
    average="macro"
)

baseline_f1 = f1_score(
    y_test,
    y_pred,
    average="macro"
)

print("Baseline Accuracy        :", baseline_accuracy)
print("Baseline Macro Precision :", baseline_precision)
print("Baseline Macro Recall    :", baseline_recall)
print("Baseline Macro F1        :", baseline_f1)
```

```python
# Display detailed classification report

print(
    classification_report(
        y_test,
        y_pred,
        labels=[1, 2, 3],
        target_names=[
            "0-1 Doctors",
            "2-3 Doctors",
            "4+ Doctors"
        ]
    )
)
```

```python
# Create baseline confusion matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[1, 2, 3]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "0-1 Doctors",
        "2-3 Doctors",
        "4+ Doctors"
    ]
)

disp.plot()

plt.title("Baseline Decision Tree Confusion Matrix")
plt.tight_layout()

plt.savefig(
    "../results/eda_visualizations/baseline_decision_tree_confusion_matrix.png",
    dpi=300
)

plt.show()
```

```python

# Define Decision Tree hyperparameters

param_grid = {
    "classifier__criterion": [
        "gini",
        "entropy",
        "log_loss"
    ],

    "classifier__max_depth": [
        3,
        5,
        8,
        None
    ],

    "classifier__min_samples_split": [
        2,
        5,
        10
    ],

    "classifier__min_samples_leaf": [
        1,
        2,
        5
    ],

    "classifier__class_weight": [
        None,
        "balanced"
    ]
}

print("Parameter grid created successfully.")
```

```python
# Create 5-fold Stratified Cross-Validation

cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-fold Stratified Cross-Validation created.")
```

```python
# Create GridSearchCV

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv_strategy,
    n_jobs=-1,
    return_train_score=True
)

print("GridSearchCV created successfully.")
```

```python
# Perform hyperparameter tuning

grid_search.fit(
    X_train,
    y_train
)

print("Hyperparameter tuning completed successfully.")
```

```python
# Display the best hyperparameters

print("Best Parameters:")
print(grid_search.best_params_)
```

```python
# Display best cross-validation Macro F1

print("Best Cross-Validation Macro F1:")
print(grid_search.best_score_)
```

```python
# Display different Decision Tree configurations

results = pd.DataFrame(
    grid_search.cv_results_
)

selected_results = results[
    [
        "param_classifier__criterion",
        "param_classifier__max_depth",
        "param_classifier__min_samples_split",
        "param_classifier__min_samples_leaf",
        "param_classifier__class_weight",
        "mean_test_score",
        "rank_test_score"
    ]
]

selected_results = selected_results.sort_values(
    "rank_test_score"
)

selected_results.head(15)
```

```python
# Get the best tuned model

best_model = grid_search.best_estimator_

print("Best tuned Decision Tree selected.")
```

```python
# Generate predictions using the best model

y_pred_best = best_model.predict(
    X_test
)

print("Final predictions generated successfully.")
```

```python
# Calculate final test-set metrics

final_accuracy = accuracy_score(
    y_test,
    y_pred_best
)

final_precision = precision_score(
    y_test,
    y_pred_best,
    average="macro"
)

final_recall = recall_score(
    y_test,
    y_pred_best,
    average="macro"
)

final_f1 = f1_score(
    y_test,
    y_pred_best,
    average="macro"
)

print("Final Decision Tree Results")
print("----------------------------")
print("Accuracy        :", final_accuracy)
print("Macro Precision :", final_precision)
print("Macro Recall    :", final_recall)
print("Macro F1        :", final_f1)
```

```python
# Final classification report

print(
    classification_report(
        y_test,
        y_pred_best,
        labels=[1, 2, 3],
        target_names=[
            "0-1 Doctors",
            "2-3 Doctors",
            "4+ Doctors"
        ]
    )
)
```

```python
# Final confusion matrix

cm_best = confusion_matrix(
    y_test,
    y_pred_best,
    labels=[1, 2, 3]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_best,
    display_labels=[
        "0-1 Doctors",
        "2-3 Doctors",
        "4+ Doctors"
    ]
)

disp.plot()

plt.title("Tuned Decision Tree Confusion Matrix")
plt.tight_layout()

plt.savefig(
    "../results/eda_visualizations/tuned_decision_tree_confusion_matrix.png",
    dpi=300
)

plt.show()
```

```python
# Evaluate the best model using 5-fold Stratified Cross-Validation

cv_results = cross_validate(
    best_model,
    X_train,
    y_train,
    cv=cv_strategy,
    scoring=[
        "accuracy",
        "precision_macro",
        "recall_macro",
        "f1_macro"
    ],
    n_jobs=-1
)

print("5-Fold Cross-Validation Results")
print("--------------------------------")

print(
    "Mean Accuracy:",
    cv_results["test_accuracy"].mean()
)

print(
    "Mean Macro Precision:",
    cv_results["test_precision_macro"].mean()
)

print(
    "Mean Macro Recall:",
    cv_results["test_recall_macro"].mean()
)

print(
    "Mean Macro F1:",
    cv_results["test_f1_macro"].mean()
)
```

```python
# Display performance for each fold

cv_table = pd.DataFrame({
    "Fold": [1, 2, 3, 4, 5],

    "Accuracy":
        cv_results["test_accuracy"],

    "Macro Precision":
        cv_results["test_precision_macro"],

    "Macro Recall":
        cv_results["test_recall_macro"],

    "Macro F1":
        cv_results["test_f1_macro"]
})

cv_table
```

```python
# Extract feature importance from the best Decision Tree

trained_preprocessor = (
    best_model.named_steps["preprocessor"]
)

trained_tree = (
    best_model.named_steps["classifier"]
)

feature_names = (
    trained_preprocessor.get_feature_names_out()
)

feature_importances = pd.DataFrame({
    "Feature": feature_names,
    "Importance": trained_tree.feature_importances_
})

feature_importances = feature_importances.sort_values(
    "Importance",
    ascending=False
)

feature_importances.head(15)
```

```python
# Plot the top 15 feature importances

top_features = feature_importances.head(15)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.title("Top 15 Decision Tree Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()

plt.savefig(
    "../results/eda_visualizations/decision_tree_feature_importance.png",
    dpi=300
)

plt.show()
```

```python
# Save final Decision Tree results

model_results = pd.DataFrame({
    "Model": ["Decision Tree"],

    "Accuracy": [final_accuracy],

    "Macro Precision": [final_precision],

    "Macro Recall": [final_recall],

    "Macro F1": [final_f1],

    "CV Macro F1": [
        cv_results["test_f1_macro"].mean()
    ]
})

model_results.to_csv(
    "../results/outputs/decision_tree_results.csv",
    index=False
)

print("Decision Tree results saved successfully.")
```
